```python
import matplotlib.pyplot as plt
import numpy as np

# Data
metrics = ["Precision", "Recall", "F1"]
typical = [81.0, 57.0, 66.0]
atypical = [76.0, 55.0, 63.0]

x = np.arange(len(metrics))
width = 0.4

fig, ax = plt.subplots(figsize=(6, 5))

bars1 = ax.bar(x - width/2, typical, width, label="Typical", color="blue")
bars2 = ax.bar(x + width/2, atypical, width, label="Atypical", color="red")

ax.set_title("Overall Metrics: Typical (Blue) vs Atypical (Red)")
ax.set_ylabel("Score")
ax.set_xticks(x)
ax.set_xticklabels(metrics, rotation=15, ha="right")
ax.set_ylim(0, 100.0)
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.legend()

# Add value labels
def add_labels(bars):
    for b in bars:
        h = b.get_height()
        ax.text(b.get_x() + b.get_width()/2, h + 0.015, f"{h:.2f}",
                ha="center", va="bottom", fontsize=14)

add_labels(bars1)
add_labels(bars2)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
scp_labels = ["scp-base", "scp-1x-oversample", "scp-2x-oversample"]
scp_values = [6.7, 5.5, 5.1]
scp_colors = ["#0A0C7A", "#4B6BD6", "#BFDDE8"]

xpaste_labels = ["xpaste-base", "xpaste-1x-oversample", "xpaste-2x-oversample"]
xpaste_values = [6.3, 5.3, 5.0]
xpaste_colors = ["#F5FF5A", "#E8E59A", "#FBFBE4"]

def style_barplot(ax, title, labels, values, colors):
    x = np.arange(len(labels))
    bars = ax.bar(x, values, color=colors, width=0.72)

    ax.set_title(title, fontsize=14, weight="bold", pad=14)

    ax.set_ylim(0, 7)
    ax.set_yticks(np.arange(0, 8, 1))
    ax.tick_params(axis="y", labelsize=12, colors="#777777")

    ax.set_xticks(x)
    ax.set_xticklabels(labels, rotation=15, ha="right", fontsize=11, color="#777777")

    ax.yaxis.grid(True, linestyle="--", linewidth=1, alpha=0.35)
    ax.set_axisbelow(True)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color("#DDDDDD")
    ax.spines["bottom"].set_color("#DDDDDD")

    for b, v in zip(bars, values):
        ax.text(
            b.get_x() + b.get_width() / 2,
            v + 0.08,
            f"{v:.1f}",
            ha="center",
            va="bottom",
            fontsize=22,
            color="black",
        )

fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=True)

style_barplot(axes[0], "SCP (delta = typical - atypical)", scp_labels, scp_values, scp_colors)
style_barplot(axes[1], "XPaste (delta = typical - atypical)", xpaste_labels, xpaste_values, xpaste_colors)

plt.tight_layout()
plt.show()
```
